```python
import yfinance as yf
import pandas as pd

# Define the list of XLRE tickers
tickers = ['PLD', 'AMT', 'EQIX', 'CCI', 'PSA', 'O', 'SPG', 'DLR', 'WELL', 'AVB',
           'EXR', 'EQR', 'VTR', 'SBAC', 'ESS', 'MAA', 'ARE', 'INVH', 'UDR', 'SUI',
           'IRM', 'OHI', 'REG', 'KIM', 'VICI', 'BXP', 'WY', 'HST', 'EQC', 'DOC']

# Download 6 months of daily adjusted closing prices
# Accessing 'Adj Close' using a tuple key (assuming it's a multi-level column index)
# data = yf.download(tickers, start='2024-06-01', end='2024-12-31')[('Adj Close',)]
# If the above line raises an error, try using 'Close' instead:
data = yf.download(tickers, start='2024-06-01', end='2024-12-31')['Close']

# Display the first few rows of the data
print(data.head())

# Save the data to a CSV file for later use
data.to_csv('XLRE_Historical_Price_Data.csv')
print("\nData saved as 'XLRE_Historical_Price_Data.csv'")
```

```python
# Compute daily returns
returns = data.pct_change().dropna()

# Display the first few rows of daily returns
print(returns.head())

# Save the daily returns to a CSV file for later use
returns.to_csv('XLRE_Daily_Returns.csv')
print("\nDaily returns saved as 'XLRE_Daily_Returns.csv'")
```

```python
# Compute the covariance matrix
cov_matrix = returns.cov()

# Display the first 5 rows and 5 columns of the covariance matrix
print(cov_matrix.iloc[:5, :5])

# Save the covariance matrix to a CSV file
cov_matrix.to_csv('XLRE_Covariance_Matrix.csv')
print("\nCovariance matrix saved as 'XLRE_Covariance_Matrix.csv'")
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import pandas as pd

# Perform PCA on the covariance matrix
pca = PCA()
pca.fit(cov_matrix)

# Explained variance ratio (percentage of variance explained by each component)
explained_variance = pca.explained_variance_ratio_ * 100  # Convert to percentage

# Display explained variance for the first 5 principal components
for i, variance in enumerate(explained_variance[:5]):
    print(f"Principal Component {i+1}: {variance:.2f}% variance explained")

# Save PCA results to a CSV file
pca_results = pd.DataFrame({
    'Principal Component': [f"PC{i+1}" for i in range(len(explained_variance))],
    'Explained Variance (%)': explained_variance.round(2)
})
pca_results.to_csv('XLRE_PCA_Results.csv', index=False)
print("\nPCA results saved as 'XLRE_PCA_Results.csv'")

# Scree plot for variance explained
plt.figure(figsize=(8, 6))
plt.bar(range(1, len(explained_variance) + 1), explained_variance, color='skyblue')
plt.title('Scree Plot: Variance Explained by Principal Components')
plt.xlabel('Principal Component')
plt.ylabel('Variance Explained (%)')
plt.grid(True)
plt.show()
```

### Empirical Analysis of XLRE ETF
### Introduction
This analysis investigates the XLRE ETF, which represents the U.S. real estate sector. By analyzing the ETF’s top 30 holdings over six months, we computed daily returns, the covariance matrix, Principal Component Analysis (PCA), and Singular Value Decomposition (SVD). These techniques helped uncover patterns in the data and identify key drivers of variability among the ETF's holdings.

### Importance of Daily Returns
Daily returns measure the percentage change in asset prices from one day to the next and serve as a foundation for financial analysis. They provide insights into the volatility, risk, and performance of individual assets. For XLRE, analyzing daily returns revealed how the 30 holdings interact and how they contribute to the ETF's overall behavior. These returns also enabled the construction of the covariance matrix, which quantifies the relationships between the assets, forming the basis for advanced analyses like PCA and SVD.

### Principal Component Analysis (PCA)
PCA reduces the complexity of the covariance matrix by identifying principal components that explain the most variance in the data. Each principal component has an eigenvalue, representing the variance it explains, and an eigenvector, showing how much each stock contributes to the component.

For XLRE, the first principal component explained 45.02% of the total variance, followed by 18.40%, 14.85%, 7.68%, and 4.03% for the next four components. This means the first principal component alone captures nearly half of the variance, indicating that the performance of a few dominant stocks (such as Prologis Inc. and American Tower Corporation) significantly influences the ETF. The scree plot revealed a steep decline in explained variance after the first few components, suggesting that most of the dataset's variability can be summarized with only a few principal components.

### Singular Value Decomposition (SVD)
### Singular Value Decomposition (SVD)

SVD decomposes the returns matrix into three components:

$$A = U \Sigma V^T$$

- **$U$**: Left singular vectors (patterns across rows).
- **$\Sigma$**: Singular values (diagonal matrix showing variance magnitudes).
- **$V^T$**: Right singular vectors (patterns across columns).

The singular values in $\Sigma$ are directly related to the variance captured in each direction, similar to the eigenvalues in PCA.


For XLRE, the first five singular values were 0.6157, 0.2644, 0.2482, 0.2114, and 0.1778, with the first singular value being significantly larger. This aligns with PCA findings, where the first component explained the majority of variance. SVD confirms that XLRE's returns are dominated by a few major patterns, reinforcing the influence of key stocks.

### Comparison: PCA vs. SVD
Both PCA and SVD are dimensionality reduction techniques, but they serve different purposes:

PCA focuses on the covariance matrix, emphasizing variance and correlations. It identifies eigenvalues and eigenvectors, ideal for analyzing relationships between variables.

SVD works directly on the returns matrix, decomposing it into singular values and vectors, making it more versatile for general data decomposition.
For XLRE, PCA highlighted the variance explained by principal components, while SVD confirmed the dominance of key patterns. Both techniques revealed that XLRE’s performance is concentrated in a small subset of its holdings.


### Insights from XLRE
The analysis demonstrated that XLRE’s returns are highly influenced by a few principal components and singular values. This suggests that while the ETF is diversified, its performance heavily depends on its largest holdings. Investors should monitor these dominant stocks to better understand XLRE’s behavior.

### Conclusion
PCA and SVD provided complementary insights into XLRE's returns, simplifying complex data and identifying key drivers of variance. These findings offer actionable insights for portfolio optimization and risk management, emphasizing the ETF's reliance on its largest holdings.
